```python
cols = ['Company Names', 'Cars Names', 'Engines', 'CC/Battery Capacity', 'Total Speed', 'Performance(0 - 100 )KM/H', 'Cars Prices']

cars = df[cols].copy()

print(cars.shape)
cars.head()
```

```python
cars = cars.rename(columns={
    "Company Names": "company",
    "Cars Names": "car",
    "Engines": "engine",
    "CC/Battery Capacity": "capacity",
    "Total Speed": "top_speed",
    "Performance(0 - 100 )KM/H": "acc_0_100",
    "Cars Prices": "price"
})
cars.columns
```

```python
import re
import numpy as np

def avg_number(text):
    text = str(text).replace(",", "")           # "12,000" -> "12000"
    nums = re.findall(r"\d+\.?\d*", text)       # find every number in the text
    if not nums:                                # no numbers found (like "-")
        return np.nan
    return sum(float(n) for n in nums) / len(nums)   # average them
```

```python
cars["price_usd"]     = cars["price"].apply(avg_number)
cars["top_speed_kmh"] = cars["top_speed"].apply(avg_number)
cars["acc_sec"]       = cars["acc_0_100"].apply(avg_number)
```

```python
is_cc  = cars["capacity"].str.contains("cc",  case=False, na=False)
is_kwh = cars["capacity"].str.contains("kwh", case=False, na=False)

cars["engine_cc"]   = np.where(is_cc,  cars["capacity"].apply(avg_number), np.nan)
cars["battery_kwh"] = np.where(is_kwh, cars["capacity"].apply(avg_number), np.nan)
```

```python
cars[["price_usd", "top_speed_kmh", "acc_sec", "engine_cc", "battery_kwh"]].info()
print(cars[["price_usd", "top_speed_kmh", "acc_sec", "engine_cc", "battery_kwh"]].describe())

# Rows with neither cc nor kWh (the "-" cases) and rows with both
print((~is_cc & ~is_kwh).sum(), (is_cc & is_kwh).sum())
print(cars[~is_cc & ~is_kwh]["capacity"].unique())
```

```python
both = is_cc & is_kwh
print(cars.loc[both, ["company", "car", "capacity"]])
print(cars.loc[cars["battery_kwh"] > 250, ["company", "car", "capacity", "battery_kwh"]])
print(cars.nlargest(5, "price_usd")[["company", "car", "price", "price_usd"]])
print(cars.nlargest(5, "engine_cc")[["company", "car", "capacity", "engine_cc"]])
print(cars.nsmallest(5, "battery_kwh")[["company", "car", "engine", "capacity", "battery_kwh"]])
print(cars.nlargest(5, "acc_sec")[["company", "car", "acc_0_100", "acc_sec"]])
```

```python
def pick(text, unit):
    m = re.search(r"([\d,]*\.?\d+)\s*" + unit, str(text), flags=re.IGNORECASE)
    return float(m.group(1).replace(",", "")) if m else np.nan

cars.loc[both, "engine_cc"]   = cars.loc[both, "capacity"].apply(lambda t: pick(t, "cc"))
cars.loc[both, "battery_kwh"] = cars.loc[both, "capacity"].apply(lambda t: pick(t, "kwh"))

print(cars.loc[both, ["capacity", "engine_cc", "battery_kwh"]])
```
